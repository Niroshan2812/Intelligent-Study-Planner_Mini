```python
from google.colab import files
import io
import pandas as pd

uploaded = files.upload()

filename = next(iter(uploaded))

data = pd.read_csv(io.BytesIO(uploaded[filename]))

print(f"Success loaded {len(data)} from '{filename}'")
print(data.head())
```

```python
import numpy as np
import xgboost as xgb
from onnxmltools import convert_xgboost
from onnxmltools.convert.common.data_types import FloatTensorType
from google.colab import files

# Select features
feature_columns = ['difficulty_level', 'current_score','english_fluency','tuition_hours_weekly', 'commute_fatigue']

target_column = 'hours_needed'

X = data[feature_columns]
y = data[target_column]

# Train XGBoost
print("Training model .....")
# Explicitly set base_score to avoid conversion errors with onnxmltools
model = xgb.XGBRegressor(n_estimators=100, max_depth =4, base_score=0.5)
model.fit(X.to_numpy(), y)
print("Training Complete")

print("Convert into ONNX")

initial_type = [('float_input', FloatTensorType([None, len(feature_columns)]))]
onx = convert_xgboost(model, initial_types=initial_type)

#Save and download
outputFile = "Study_predictor.onnx"
with open(outputFile, "wb") as f:
    f.write(onx.SerializeToString())

print(f"Model saved to '{outputFile}'")
files.download(outputFile)
```
